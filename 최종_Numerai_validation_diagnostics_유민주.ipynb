```python
!pip install numerapi -q

import numerapi
import getpass

NUMERAI_PUBLIC_ID = getpass.getpass("Public ID 입력: ").strip()
NUMERAI_SECRET_KEY = getpass.getpass("Secret Key 입력: ").strip()

napi = numerapi.NumerAPI(public_id=NUMERAI_PUBLIC_ID, secret_key=NUMERAI_SECRET_KEY)
print(napi.get_models())
```

## Ensemble1: XGBoost CatBoost LightGBM ExtraTrees RandomForest

```python
models = napi.get_models()

model_id_ensemble1 = models["ensemble1_5models"]

status_ensemble1 = napi.diagnostics(
    model_id=model_id_ensemble1)

print(type(status_ensemble1))
print(status_ensemble1)
```

```python
key_fields = [
    "status",
    "validationCorrSharpe",
    "validationCorrMean",
    "validationCorrStd",
    "validationFeatureNeutralCorrV3Mean",
    "validationBmcMean",
    "validationFeatureCorrMax",
    "validationMaxDrawdown",
    "validationAutocorr",
    "validationCorrCorrWExamplePreds",
]

for item in status_ensemble1:
    for k in key_fields:
        if k in item:
            print(f"{k}: {item[k]}")
```

```python
import pandas as pd

diagnostics_df = pd.DataFrame(status_ensemble1)
diagnostics_df = diagnostics_df.dropna(axis=1, how="all")
display(diagnostics_df)
```
