## **4.1 인공 신경망의 한계와 딥러닝 출현**

인공 신경망에서 이용하는 구조(입력층, 출력층, 가중치로 구성된 구조)는 프랭크 로젠블라트가 1957년에 고안한 퍼셉트론이라는 선형 분류기.

**퍼셉트론**

- 오늘날 신경망(딥러닝)의 기원이 되는 알고리즘
- 다수의 신호(흐름이 있는)를 입력으로 받아 하나의 신호를 출력함. 이 신호를 입력으로 받아 '흐른다/안 흐른다(1 또는 0)'는 정보를 앞으로 전달하는 원리로 작동함

#### **AND 게이트**
- 모든 입력이 ‘1’일 때 작동
- 즉, 입력 중 어떤 하나라도 ‘0’을 가지면 작동을 멈춤

#### **OR 게이트**
- 입력에서 둘 중 하나만 ‘1’이거나 둘 다 ‘1’일 때 작동
- 즉, 입력 모두가 ‘0’인 경우를 제외한 나머지는 모두 ‘1’ 값을 가짐

#### **XOR 게이트**
- 배타적 논리합이라고도 하며, 입력 두 개 중 한 개만 ‘1’일 때 작동하는 논리 연산. 데이터가 비선형적으로 분리되기 때문에 제대로 분류하기 어려움.
- 즉, 단층 퍼셉트론은 AND, OR 연산은 학습할 수 있지만 XOR은 학습할 수 없음
이를 극복하는 방안으로 다층 퍼셉트론을 고안함

#### **다층 퍼셉트론**
- 입력층과 출력층 사이에 하나 이상의 중간층(은닉층)을 두어 비선형적으로 분리되는 데이터도 학습할 수 있음
- 심층 신경망(DNN): 입력층과 출력층 사이에 은닉층이 여러 개 있는 신경망. 다른 이름으로 딥러닝이라고 함

## **4.2 딥러닝 구조**
딥러닝은 여러 층을 가진 인공 신경망을 사용해 학습을 수행하는 것. 입력층, 출력층과 두 개 이상의 은닉층으로 구성됨.
### **4.2.1 딥러닝 용어**
- 입력층: 데이터를 받아들이는 층
- 은닉층: 모든 입력 노드로부터 입력값을 받아 가중합을 계산하고, 이 값을 활성화 함수에 적용해 출력층에 전달하는 층
- 출력층: 신경망의 최종 결괏값이 포함된 층
- 가중치(weight): 노드와 노드 사이의 연결 강도
- 바이어스(bias): 가중합에 더해 주는 상수. 하나의 뉴런에서 활성화 함수를 거쳐 최종적으로 출력되는 값을 조절하는 역할
- 가중합(weighted sum), 전달 함수: 가중치와 신호를 곱한 값들의 합
- 활성화 함수: 신호를 입력받아 적절히 처리한 뒤 출력해 주는 함수
- 손실 함수: 가중치 학습을 위해 출력 함수의 결과와 실젯값 사이의 오차를 측정하는 함수

#### **가중치**
- 입력값이 연산 결과에 미치는 영향력을 조절하는 요소
- 예) w₁ 값이 0이거나 0에 가까운 0.001이면, x₁이 아무리 커도 x₁ × w₁ 값은 0이거나 0에 가까워짐

image.png


#### **가중합 또는 전달 함수**
- 각 노드에서 들어오는 신호에 가중치를 곱해 다음 노드로 전달하는데, 이 값들을 모두 더한 합계가 가중합
- 노드의 가중합이 계산되면 이 값을 활성화 함수로 보내기 때문에 전달 함수라고도 함
공식: Σ wᵢxᵢ + b (w: 가중치, b: 바이어스)

#### **활성화 함수**
- 전달 함수에서 받은 값을 출력할 때 일정 기준에 따라 출력값을 바꾸는 비선형 함수(직선으로 표현할 수 없는 데이터 사이의 관계를 표현하는 함수)
- 종류: 시그모이드, 하이퍼볼릭 탄젠트, 렐루, 리키 렐루, 소프트맥스 등

- **시그모이드 함수**
    - 선형 함수의 결과를 0~1 사이의 비선형 형태로 변형
    - 주로 로지스틱 회귀처럼 분류 문제를 확률적으로 표현하는 데 사용
    - 모델이 깊어지면 기울기가 사라지는 기울기 소멸 문제가 생겨 딥러닝 모델에서는 잘 사용하지 않음
    - 수식: $f(x) = 1 / (1 + e⁻ˣ)$
    - 기울기 소멸 문제: 활성화 함수 계산 과정에서 도함수 값이 계속 곱해지면서 가중치 결괏값이 0이 되어 더 이상 계산할 수 없는 상태
- **하이퍼볼릭 탄젠트 함수**
    - 선형 함수의 결과를 -1~1 사이의 비선형 형태로 변형
    - 시그모이드에서 결괏값의 평균이 0이 아닌 양수로 편향되는 문제를 해결하는 데 사용함
    - 하지만 기울기 소멸 문제는 여전히 발생
- **렐루(ReLU) 함수**
    - 입력(x)이 음수면 0을, 양수면 x를 출력
    - 경사 하강법에 영향을 주지 않아 학습 속도가 빠르고 기울기 소멸 문제가 발생하지 않음
    - 일반적으로 은닉층에서 사용하며, 하이퍼볼릭 탄젠트 함수보다 학습 속도가 6배 빠름
    - 문제점: 음수 값을 입력받으면 항상 0을 출력해 학습 능력이 떨어짐 → 리키 렐루 등을 사용해 해결
- **리키 렐루(Leaky ReLU) 함수**
    - 입력값이 음수면 0이 아닌 0.001처럼 매우 작은 수를 반환
    - 입력값이 수렴하는 구간이 사라져 렐루 함수의 문제를 해결함
- **소프트맥스(softmax) 함수**
    - 입력값을 0~1 사이로 정규화해 출력값들의 총합이 항상 1이 되도록 함
    - 보통 딥러닝에서 출력 노드의 활성화 함수로 많이 사용
    - 수식: $yₖ = exp(aₖ) / Σᵢ₌₁ⁿ exp(aᵢ)$
        - 분자는 입력 신호 aₖ의 지수 함수, 분모는 모든 입력 신호의 지수 함수 합
        - n은 출력층의 뉴런 개수, yₖ는 그중 k번째 출력

```python
import torch

class Net(torch.nn.Module):
    def __init__(self, n_feature, n_hidden, n_output):
        super(Net, self).__init__()
        self.hidden = torch.nn.Linear(n_feature, n_hidden) # 은닉층
        self.relu = torch.nn.ReLU(inplace=True)
        self.out = torch.nn.Linear(n_hidden, n_output) # 출력층
        self.softmax = torch.nn.Softmax(dim=n_output)
    def forward(self, x):
        x = self.hidden(x)
        x = self.relu(x) # 은닉층을 위한 렐루 활성화 함수
        x = self.out(x)
        x = self.softmax(x) # 출력층을 위한 소프트맥스 활성화 함수
        return x
```

#### **손실 함수**
경사 하강법: 학습률(η)과 손실 함수의 순간 기울기를 이용해 가중치를 업데이트하는 방법. 즉, 미분의 기울기로 오차를 비교하고 오차를 최소화하는 방향으로 이동시키는 방법
- 학습률: 한 번 학습할 때 얼마큼 변화를 줄지 정하는 상수

손실 함수는 이때 오차를 구하는 방법이며, 학습으로 얻은 추정치가 실제 데이터와 얼마나 차이 나는지 평가하는 지표
- 값이 클수록 많이 틀렸다는 의미, ‘0’에 가까울수록 완벽하게 추정했다는 의미
- 대표적인 손실 함수: 평균 제곱 오차(MSE), 크로스 엔트로피 오차(CEE)
- **평균 제곱 오차(MSE)**
    - 실젯값과 예측값의 차이(error)를 제곱해 평균을 낸 것
    - 값이 작을수록 예측력이 좋다는 의미

    - 수식: $MSE = (1/n) Σ(yᵢ − ŷᵢ)²$

### **4.2.2 딥러닝 학습**
딥러닝 학습은 크게 순전파와 역전파 두 단계로 진행됨

image.png


**순전파(feedforward)**
- 네트워크에 훈련 데이터가 들어올 때 일어나며, 예측값을 계산하기 위해 전체 신경망을 지나감
- 모든 뉴런이 이전 층의 뉴런에서 받은 정보에 변환(가중합 및 활성화 함수)을 적용해 다음 층(은닉층)의 뉴런으로 전송함
- 데이터가 모든 층을 통과하고 모든 뉴런의 계산이 끝나면 예측값이 최종 층(출력층)에 도달함

**손실 계산**
- 손실 함수로 네트워크의 예측값과 실젯값의 차이(손실 오차)를 추정함
- 손실 함수 비용은 ‘0’이 이상적이므로, 비용이 0에 가까워지도록 훈련을 반복하며 가중치를 조정함

**역전파(backpropagation)**
- 계산된 손실(오차) 정보를 역으로(출력층 → 은닉층 → 입력층) 전파함
- 은닉층의 뉴런은 각 뉴런이 원래 출력에 기여한 상대적 기여도(가중치)에 따라 값이 달라짐
- 수학적으로는 예측값과 실젯값의 차이를 각 뉴런의 가중치로 미분한 뒤 기존 가중치에서 뺌
- 이 과정을 출력층 → 은닉층 → 입력층 순서로 모든 뉴런에 진행하고, 계산된 결과를 다시 순전파의 가중치로 사용함

#### **경사 하강법의 유형**
image.png


- **배치 경사 하강법(BGD)**
    - 전체 데이터셋의 오류를 구한 뒤 기울기를 한 번만 계산해 모델의 파라미터를 업데이트하는 방법
    - 즉, 전체 훈련 데이터셋에 대해 가중치를 편미분함
    - 수식: $W = W − α∇J(W, b) (α: 학습률, J: 손실 함수)$
    - 단점: 한 스텝에 모든 훈련 데이터를 사용하므로 학습이 오래 걸림
- **확률적 경사 하강법(SGD)**
    - 무작위로 고른 데이터에 대해 기울기를 계산하는 방법
    - 적은 데이터를 사용하므로 계산이 빠름
    - 파라미터 변경 폭이 불안정하고 때로는 배치 경사 하강법보다 정확도가 낮을 수 있지만, 속도가 빠름
- **미니 배치 경사 하강법**
    - 전체 데이터셋을 여러 개의 미니 배치로 나누고, 미니 배치마다 기울기를 구한 뒤 그 평균 기울기로 모델을 업데이트함


#### **옵티마이저(optimizer)**
확률적 경사 하강법의 파라미터 변경 폭이 불안정한 문제를 해결하기 위해, 학습 속도와 운동량을 조정하는 옵티마이저를 적용할 수 있음

· **속도를 조정하는 방법**
- **아다그라드(Adagrad)**
    - 변수(가중치)의 업데이트 횟수에 따라 학습률을 조정하는 방법
    - 많이 변한 변수는 최적값에 가까워졌다고 보고 학습률을 작게 해 세밀하게 조정함
    - 적게 변한 변수는 학습률을 크게 해 빠르게 오차를 줄임
    - G 함수(기울기 크기의 누적)를 추가해 파라미터마다 다른 학습률을 줌 → 기울기가 크면 G가 커지고 학습률은 작아짐
    - 단점: 기울기가 0에 수렴하는 문제가 있어 잘 사용하지 않고, 대신 RMSProp을 사용
- **아다델타(Adadelta)**
    - 아다그라드에서 G 값이 커지면서 학습이 멈추는 문제를 해결하기 위해 등장
    - 학습률($η$)을 D 함수(가중치 변화량 크기의 누적값)로 바꿨기 때문에 학습률 하이퍼파라미터가 필요 없음
- **RMSProp**
    - 아다그라드의 G(i) 값이 무한히 커지는 것을 막기 위해 제안됨
    - G 함수에 γ(감마)만 추가함 → G가 너무 커서 학습률이 작아지는 것을 막기 위해 γ로 학습률 크기를 비율로 조정할 수 있게 함

· **운동량을 조정하는 방법**
- **모멘텀(Momentum)**
    - 매번 기울기를 구하지만, 가중치를 수정하기 전에 이전 수정 방향(+, −)을 참고해 같은 방향으로 일정 비율만 수정함
    - +/− 방향으로 번갈아 수정되는 지그재그 현상이 줄고, 이전 이동값을 고려하므로 관성 효과를 얻을 수 있음
    - SGD와 함께 사용(SGD 모멘텀): 기울기를 속도(v)로 대체해 이전 속도의 일정 부분을 반영함
    - momentum 값은 0.9에서 시작해 0.95, 0.99처럼 조금씩 늘리며 사용
- **네스테로프 모멘텀(NAG)**
    - 모멘텀 값이 적용된 지점에서 기울기를 계산함
    - 모멘텀은 멈춰야 할 시점에도 관성 때문에 훨씬 멀리 갈 수 있다는 단점이 있음
    - 네스테로프는 모멘텀으로 절반 정도 이동한 뒤 어떻게 이동할지 다시 계산해 결정하므로, 빠른 이동 속도는 유지하면서 적절한 시점에 제동을 걸기 쉬움

· **속도와 운동량을 함께 쓰는 방법**
- **아담(Adam)**
    - 모멘텀과 RMSProp의 장점을 결합한 경사 하강법
    - RMSProp의 G 함수(기울기 제곱의 지수 평균)와 모멘텀의 v(i)를 사용해 가중치를 업데이트함
    - 현재 가장 보편적이며 성능도 우수함

### **4.2.4 딥러닝을 사용할 때 이점**
#### **특성 추출**
- 특성 추출(feature extraction): 데이터별로 어떤 특징이 있는지 찾아내고, 그것을 토대로 데이터를 벡터로 바꾸는 작업
- 사람이 인지하는 데이터를 컴퓨터가 인지할 수 있는 데이터로 바꿔야 패턴이나 규칙을 찾을 수 있음
- SVM, 나이브 베이즈, 로지스틱 회귀 같은 머신 러닝 알고리즘은 특성 추출이 매우 복잡하고 데이터 분야에 대한 전문 지식이 필요했음
- 딥러닝은 특성 추출 과정을 알고리즘에 통합함 → 은닉층을 깊게 쌓아 파라미터를 늘린 모델 구조 덕분

#### **빅데이터의 효율적 활용**
- 딥러닝이 특성 추출을 알고리즘에 통합할 수 있는 이유는 빅데이터 때문
- 딥러닝을 이용한 특성 추출은 데이터가 많을수록 성능이 좋아짐
- 즉, 확보한 데이터가 적다면 딥러닝의 성능 향상을 기대하기 어려우므로 머신 러닝을 고려해야 함

## **4.3 딥러닝 알고리즘**
딥러닝 알고리즘은 모두 심층 신경망을 사용한다는 공통점이 있음. 목적에 따라 합성곱 신경망(CNN), 순환 신경망(RNN), 제한된 볼츠만 머신(RBM), 심층 신뢰 신경망(DBN)으로 나뉨.

### **4.3.1 심층 신경망(DNN)**
- 입력층과 출력층 사이에 여러 은닉층을 포함하는 인공 신경망
- 머신 러닝은 비선형 분류를 위해 여러 트릭(trick)을 사용했지만, DNN은 은닉층을 여러 개 추가했기 때문에 별도의 트릭 없이 비선형 분류가 가능함
- 장점: 다양한 비선형 관계를 학습할 수 있음
- 단점: 학습 연산량이 많고 기울기 소멸 문제 등이 생길 수 있음 → 드롭아웃, 렐루 함수, 배치 정규화 등을 적용해야 함

### **4.3.2 합성곱 신경망(CNN)**
- 합성곱층과 풀링층을 포함하는, 이미지 처리 성능이 좋은 인공 신경망 알고리즘
- 영상·사진 같은 이미지 데이터에서 객체를 탐색하거나 객체 위치를 찾는 데 유용함
- 이미지에서 객체, 얼굴, 장면을 인식하기 위해 패턴을 찾는 데 특히 유용함
- 대표 모델: LeNet-5, AlexNet / 층을 더 깊게 쌓은 모델: VGG, GoogLeNet, ResNet
- 기존 신경망과의 차별점
    - 각 층의 입출력 형상을 유지함
    - 이미지의 공간 정보를 유지하면서 인접 이미지와 차이가 있는 특징을 효과적으로 인식함
    - 여러 필터로 이미지의 특징을 추출하고 학습함
    - 추출한 이미지의 특징을 모으고 강화하는 풀링층이 있음
    - 필터를 공유 파라미터로 사용하므로 일반 인공 신경망보다 학습 파라미터가 매우 적음

image.png


### **4.3.3 순환 신경망(RNN)**
- 시계열 데이터(음악, 영상 등)처럼 시간 흐름에 따라 변하는 데이터를 학습하기 위한 인공 신경망
- '순환(recurrent)'은 자기 자신을 참조한다는 뜻으로, 현재 결과가 이전 결과와 연관이 있다는 의미
- 특징
    - 시간성(temporal property)을 가진 데이터가 많음
    - 시간성 정보를 이용해 데이터의 특징을 잘 다룸
    - 시간에 따라 내용이 변하므로 데이터가 동적이고 길이가 가변적임
    - 매우 긴 데이터를 처리하는 연구가 활발히 진행 중
- 기울기 소멸 문제로 학습이 제대로 되지 않는 문제가 있음 → 메모리 개념을 도입한 **LSTM**을 많이 사용함
- 자연어 처리 분야와 잘 맞음: 언어 모델링, 텍스트 생성, 자동 번역(기계 번역), 음성 인식, 이미지 캡션 생성 등

image.png


### **4.3.4 제한된 볼츠만 머신(RBM)**
- 볼츠만 머신: 가시층(visible layer)과 은닉층(hidden layer)으로 구성된 모델
- 제한된 볼츠만 머신: 가시층이 은닉층과만 연결되는 모델(가시층끼리, 은닉층끼리는 연결이 없음)
- 특징
    - 차원 감소, 분류, 선형 회귀 분석, 협업 필터링, 특성 값 학습, 주제 모델링에 사용
    - 기울기 소멸 문제를 해결하기 위한 사전 학습 용도로 활용할 수 있음
    - 심층 신뢰 신경망(DBN)의 요소로 활용됨
- 딥러닝에서 많이 쓰는 알고리즘은 CNN과 RNN이고, RBM과 DBN은 상대적으로 많이 쓰지 않음

### **4.3.5 심층 신뢰 신경망(DBN)**
- 입력층과 은닉층으로 구성된 RBM을 블록처럼 여러 층으로 쌓아 연결한 신경망
- 즉, 사전 훈련된 RBM을 층층이 쌓아 올린 구조로, 레이블이 없는 데이터에 대한 비지도 학습이 가능함
- 부분적인 이미지에서 전체를 연상하는 일반화·추상화 과정을 구현할 때 유용함
- 학습 절차
    1. 가시층과 은닉층 1에 RBM을 사전 훈련함
    2. 첫 번째 층의 입력 데이터와 파라미터를 고정하고 두 번째 층의 RBM을 사전 훈련함
    3. 원하는 층 수만큼 RBM을 쌓아 올려 전체 DBN을 완성함
- 특징
    - 순차적으로 학습하면서 계층적 구조를 만듦
    - 비지도 학습으로 학습함
    - 위로 올라갈수록 추상적인 특성을 추출함
    - 학습된 가중치를 다층 퍼셉트론의 가중치 초깃값으로 사용함

## **4.4 우리는 무엇을 배워야 할까?**
- 머신 러닝과 딥러닝 중 무엇을 쓸지는 주어진 데이터로 어떤 결과를 얻고 싶은지에 따라 달라짐
- 간단한 선형 회귀·분류로 원하는 값을 얻을 수 있다면 머신 러닝으로 충분함
- 복잡한 비선형 데이터를 분류하거나 예측하려면 딥러닝을 사용해야 함
- 즉, 머신 러닝과 딥러닝은 원하는 결과를 얻기 위한 도구일 뿐이며, 어떤 도구가 더 좋은 성능을 낼지는 데이터를 수집하고 분석하는 사람이 판단해야 함
