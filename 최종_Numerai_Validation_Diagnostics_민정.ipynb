```python
!pip install numerapi -q

import numerapi
import getpass

NUMERAI_PUBLIC_ID = getpass.getpass("Public ID 입력: ").strip()
NUMERAI_SECRET_KEY = getpass.getpass("Secret Key 입력: ").strip()

napi = numerapi.NumerAPI(public_id=NUMERAI_PUBLIC_ID, secret_key=NUMERAI_SECRET_KEY)
print(napi.get_models())
```

## Ridge Elastic (Fixed Version)

https://github.com/smichelle0804/26-1-ESAA_Numerai-/blob/main/numerai_ridge_elastic_ver2.ipynb


```python
models = napi.get_models()

model_id_ridge_elastic_fixed = models["ridge_elastic_fixed"]
diagnostics_id_ridge_elastic_fixed = "e22c1217-b39f-471f-a4d9-bb08505d4e35"

status_ridge_elastic_fixed = napi.diagnostics(
    model_id=model_id_ridge_elastic_fixed,
    diagnostics_id=diagnostics_id_ridge_elastic_fixed,
)

print(type(status_ridge_elastic_fixed))
print(status_ridge_elastic_fixed)
```

```python
key_fields = [
    "status",
    "validationCorrSharpe",
    "validationCorrMean",
    "validationCorrStd",
    "validationFeatureNeutralCorrV3Mean",
    "validationBmcMean",
    "validationFeatureCorrMax",
    "validationMaxDrawdown",
    "validationAutocorr",
    "validationCorrCorrWExamplePreds",
]

for item in status_ridge_elastic_fixed:
    for k in key_fields:
        if k in item:
            print(f"{k}: {item[k]}")
```

```python
diagnostics_df = pd.DataFrame(status_ridge_elastic_fixed)
diagnostics_df = diagnostics_df.dropna(axis=1, how="all")
display(diagnostics_df)
```
